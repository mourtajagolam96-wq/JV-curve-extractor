````python
import anthropic
from dotenv import load_dotenv
import os
````

````python
load_dotenv()
api_key = os.environ.get('ANTHROPIC_API_KEY')
if api_key:
    print('API key is successfully taken')
````

````python
client = anthropic.Anthropic(api_key=api_key)
````

````python
import pandas as pd
import matplotlib.pyplot as plt

# Load and filter data
data = pd.read_csv("0min_cell_4.csv")
data = data[data["Gfrac"] == 1.0]
jv_data = data[(data["Vext"] > -0.5) & (data["Jext"] < 200)]
jv_data = jv_data.reset_index(drop=True)

# Plot and save as image
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(jv_data["Vext"], jv_data["Jext"], color="blue", s=10, label="Real Data")
ax.set_xlabel("Voltage (V)")
ax.set_ylabel("Current Density (mA/cm²)")
ax.set_title("Real J-V Curve (Gfrac = 1.0)")
ax.legend()
#ax.grid(True, alpha=0.3)

# Save to images folder
plt.savefig("image/jv_real.png", dpi=150, bbox_inches="tight")
plt.show()
````

````python
import base64
from pathlib import Path

image_path = Path('image/jv_real.png')
if image_path.exists():
    print('Image found')
else:
    print('Image not found')
````

````python
# Read the image as raw bytes
image_bytes = image_path.read_bytes()
# Convert raw bytes to base64 string
base64_image = base64.b64encode(image_bytes).decode("utf-8")
print("Type:", type(base64_image))
print("First 50 characters:", base64_image[:50])
````

````python
# Send image to Claude with a prompt
response = client.messages.create(
    model="claude-opus-4-5",
    #model="claude-sonnet-4-6",
    max_tokens=4000,
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "image",
                    "source": {
                        "type": "base64",
                        "media_type": "image/png",
                        "data": base64_image,
                    },
                },
                {
                    "type": "text",
                    "text": """You are an expert in photovoltaic J-V curve digitization.
                    Analyze the attached J-V curve image and extract numerical voltage-current density data as accurately as possible.
                    

                    1. Use VLM
                    2. Detect plot area of this image
                    3. Detect x-axis and y-axis and units of them
                    4. OCR tick labels
                    5. Detect maximum power point
                    6. calibrate pixel coordinates to real V and J values
                    7. extract curve pixels
                    8. Extra sampling is added around the knee and less number of points region
                    9. convert pixels to numerical V-J data
                    10. Do not invent data outside the visible plot range.
            
                    
                    Return your answer in this exact JSON format:
                    {
                        "data_points": [
                            {"V": 0.0, "J": 0.0},
                            {"V": 0.1, "J": 0.0}
                        ],
                        
                    }
                    Return JSON only. No extra text."""
                }
            ],
        }
    ],
)

print(response.content[0].text)
````

````python
# Send image to Claude with a prompt
response = client.messages.create(
    model="claude-opus-4-5",
    #model="claude-sonnet-4-6",
    max_tokens=16000,
    thinking={
        "type": "enabled",
        "budget_tokens": 10000  # how many tokens Claude can use for thinking
    },
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "image",
                    "source": {
                        "type": "base64",
                        "media_type": "image/png",
                        "data": base64_image,
                    },
                },
                {
                    "type": "text",
                    "text": """You are an expert in photovoltaic J-V curve digitization.
                    Analyze the attached J-V curve image and extract numerical voltage-current density data as accurately as possible.

                    Important rule:
                    The goal is pixel-level digitization of the visible curve, not a smooth estimated J-V curve.
                    Do not invent, smooth, extrapolate, or physically fit the curve.
                    Extract only the visible plotted curve or visible data markers.

                    1. Use VLM to understand the plot type, curve, axes, labels, and units.
                    2. Detect the inner plot area of this image, excluding legend, axis labels, tick labels, and margins.
                    3. Detect x-axis and y-axis labels and their units.
                    4. OCR the visible tick labels on both axes. Use only clearly readable tick labels. Do not guess unclear tick labels.
                    5. First extract the visible J-V curve/data points.
                    6. Calibrate pixel coordinates to real V and J values using the detected plot area and tick labels.
                    7. Extract curve pixels or marker centers. Use the center of the curve line or the center of plotted markers, not the edge of the line/marker.
                    8. Add extra sampling around the knee/maximum-power-point region and near the steep Voc region. Use fewer points in the almost-flat current region.
                    9. Convert the extracted pixel coordinates to numerical V-J data.
                    10. Do not invent data outside the visible plot range.
                    11. The extracted points must visually overlap with the original curve.
                    12. Ignore grid lines, axes, text, legend, dashed reference lines, and background.
                    13. Sort all extracted data points by increasing voltage.
            
                    Return your answer in this exact JSON format:
                    {
                        "data_points": [
                            {"V": 0.0, "J": 0.0},
                            {"V": 0.1, "J": 0.0}
                        ]
                    }

                    Return JSON only. No extra text."""
                }
            ],
        }
    ],
)

#print(response.content[0].text)
````

````python
import json
import re

# Step 1: Find the text block from response
raw_text = None
for block in response.content:
    if block.type == "text":
        raw_text = block.text
        break

# Step 2: Clean and parse JSON
clean_text = re.sub(r'```json|```', '', raw_text).strip()
claude_output = json.loads(clean_text)

# Step 3: Extract data points
data_points = claude_output["data_points"]

# Step 4: Print results
print("Total data points extracted:", len(data_points))
print("\nFirst 3 data points:")
for point in data_points[:3]:
    print("  V:", point["V"], "  J:", point["J"])
````

````python
import matplotlib.pyplot as plt

# Extract V and J values into separate lists
V_values = [point["V"] for point in data_points]
J_values = [point["J"] for point in data_points]

# Create the plot
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(V_values, J_values, color="blue", s=20, label="Extracted Data")
ax.scatter(jv_data["Vext"], jv_data["Jext"], color="red", s=10, label="Real Data")

# Add vertical and horizontal lines for Voc and Jsc
ax.axhline(y=0, color='gray', linestyle='--', linewidth=1)
ax.axvline(x=0, color='gray', linestyle='--', linewidth=1)

# Mark key parameters on the plot
ax.axhline(y=-parameters["Jsc"], color='red', linestyle=':', linewidth=1.5, label=f"Jsc = {parameters['Jsc']} mA/cm²")
ax.axvline(x=parameters["Voc"], color='green', linestyle=':', linewidth=1.5, label=f"Voc = {parameters['Voc']} V")

# Labels and title
ax.set_xlabel("Voltage (V)", fontsize=13)
ax.set_ylabel("Current Density (mA/cm²)", fontsize=13)
#ax.set_title(f"Extracted J-V Curve\nPCE = {parameters['PCE']}%  |  FF = {parameters['FF']}", fontsize=14)

# Add legend
ax.legend(fontsize=11)

# Add grid
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

````python
# Calculate R2
import numpy as np
from scipy import interpolate
from sklearn.metrics import r2_score

interp_func = interpolate.interp1d(
    V_values, 
    J_values,
    kind="linear",  
    fill_value="extrapolate"  )

J_extrapolated = interp_func(jv_data["Vext"])
r2 = r2_score(jv_data["Jext"], J_extrapolated)


# Step 5: Plot
fig, ax = plt.subplots(figsize=(7, 7))

ax.scatter(jv_data["Jext"], J_extrapolated,
           color="blue", s=10, alpha=0.5, label="Data points")

# Perfect fit line y = x
min_val = min(jv_data["Jext"].min(), J_extrapolated.min())
max_val = max(jv_data["Jext"].max(), J_extrapolated.max())
ax.plot([min_val, max_val], [min_val, max_val],
        'r--', linewidth=2, label="Perfect fit (y = x)")

ax.set_xlabel("Real J (mA/cm²)", fontsize=13)
ax.set_ylabel("Extracted J (mA/cm²)", fontsize=13)
ax.set_title(f"Real vs Extracted J\nR² = {round(r2, 4)}", fontsize=14)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("R² Score:", round(r2, 4))
````
